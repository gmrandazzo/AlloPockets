## Asynchronous

Features are calculated with the scripts on `utils/features_scripts`:
- TransferEntropy must be calculated with special memory handling.
- HHBlits must be calculated with reduced parallelization.
- ProDy is already parallel so it's calculated synchronously.
- The rest are parallelized normally.

<br>

**Example**

```python
from concurrent.futures import ProcessPoolExecutor, as_completed
```

```python
with ProcessPoolExecutor(max_workers=30) as executor:
    futures = {}
    
    for pdb in representatives[:2]:
        print(pdb)
        pdbpath = f"{features_path}/features/{pdb}"
        
        for fc in FClasses:
            file = f"{pdbpath}/{fc.__name__}.pkl"
            if not os.path.isfile(file):
                os.makedirs(pdbpath, exist_ok=True)
                future = executor.submit(calculate_features, pdb, fc, file, structures_path, original_cifs_path)
                futures[future] = [pdb, fc.__name__]
            
    for future in tqdm(as_completed(futures), total=len(futures), smoothing=0):
        pdb, fc = futures[future]
        try:
            calculated = future.result()
            if calculated is False:
                with open(f"{features_path}/calculation_errors.txt", "a") as f:
                    f.write(f"{pdb}\t{fc}\n")
        except:
            with open(f"{features_path}/errors.txt", "a") as f:
                f.write(f"{pdb}\t{fc}\n")
```

```python
errors_pdbs = errors[0].unique().tolist()
len(errors_pdbs), errors_pdbs
```

```python
# PDBs that don't have an error but have a missing .pkl file of a feature class
missing = []

for pdb in representatives:
    if pdb not in errors_pdbs:
        files = tuple(f for f in os.listdir(f"{features_path}/features/{pdb}") if f.endswith(".pkl"))
        miss = tuple(fc.__name__ for fc in FClasses if f"{fc.__name__}.pkl" not in files)
        if len(miss):
            missing.append((pdb, miss))

missing
```

```python
errorsf = f"{features_path}/errors.pkl"

with open(errorsf, "wb") as f:
    pickle.dump(errors_pdbs, f)
```

# Processing

```python
# Generate and save dictionary of a DataFrame of all the residues, labels, and features of each PDB
featuresf = f"{features_path}/features.pkl"

if not os.path.isfile(featuresf):
    featuresd = {}
    for pdb in tqdm(representatives, smoothing=0):
        if pdb not in errors_pdbs:
            feats = get_pdb_features(
                Cif(pdb),
                sites = [
                    sd["site"][res_cols]
                    for sid, site in sites_reps[pdb].items()
                    for sd in site["equivalent"] + [{"site": site["site"]},]
                ],
                features_path = features_path
            )
            featuresd[pdb] = pd.concat(
                (
                    pd.DataFrame([pdb]*len(feats), columns=pd.MultiIndex.from_tuples([("Residues", "pdb")])),
                    feats
                ),
                axis=1
            )
    with open(featuresf, "wb") as f:
        pickle.dump(featuresd, f)
else:
    with open(featuresf, "rb") as f:
        featuresd = pickle.load(f)

len(featuresd), featuresd
```

- FreeSASA's `relative-area_side-chain` feature has a lot of NaNs due to glycines and it's not used.
- DSSP's `relative ASA` is NaN in some cases due to modified residues, e.g. 2qmx has MSE which was remapped to MET.
- Biopython is known to fail for some residues depending on their missing atoms.
